## Python for Machine Learning & Data Preprocessing

This notebook demonstrates loading, inspecting, cleaning, encoding and scaling a sample customer dataset.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv('customer_dataset_raw.csv')
print('Shape:', df.shape)
display(df.head())
```

## 1. Initial inspection


```python
print(df.info())
print('\nMissing values:')
print(df.isnull().sum())
print('\nSummary statistics:')
display(df.describe(include='all'))
```

## 2. Handling missing values
Numerical missing values are replaced with the median of the respective column.

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Annual_Income'] = df['Annual_Income'].fillna(df['Annual_Income'].median())
print(df.isnull().sum())
```

## 3. Feature selection
Customer_ID is an identifier and is removed because it does not provide a meaningful predictive feature.

```python
df = df.drop(columns=['Customer_ID'])
display(df.head())
```

## 4. Encoding categorical variables
One-hot encoding converts Gender and City into numerical indicator columns.

```python
df_encoded = pd.get_dummies(df, columns=['Gender', 'City'], dtype=int)
display(df_encoded.head())
```

## 5. Normalization
Min-Max scaling converts Age and Annual_Income to a common 0–1 range.

```python
scaler = MinMaxScaler()
df_encoded[['Age', 'Annual_Income']] = scaler.fit_transform(df_encoded[['Age', 'Annual_Income']])
display(df_encoded.head())
```

## 6. Basic exploratory analysis

```python
print('Average age:', round(df['Age'].mean(), 2))
print('\nPurchase counts:')
print(df['Purchased'].value_counts())
print('\nCorrelation of numeric variables:')
display(df[['Age', 'Annual_Income']].corr())
```

```python
df_encoded.to_csv('customer_dataset_preprocessed.csv', index=False)
print('Preprocessed dataset saved successfully.')
```
